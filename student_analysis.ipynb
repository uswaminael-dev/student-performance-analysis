```python
df[df["Study_Hours"].isnull()]
```

```python
df["Study_Hours"] = df["Study_Hours"].fillna(
    df["Study_Hours"].median()
)
```

## Data Cleaning and Preprocessing

The dataset was cleaned using Pandas.

Steps performed:

1. Identified missing values.
2. Replaced missing Study_Hours with the median.
3. Identified and removed duplicate records.
4. Checked data types.
5. Validated attendance and marks ranges.

After cleaning, the dataset contains 9 unique student
records with no missing values or duplicates.

```python
print("Average Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
print("Average Attendance:", df["Attendance"].mean())
print("Average Study Hours:", df["Study_Hours"].mean())
```

```python
top_student = df.loc[df["Marks"].idxmax()]

print(top_student)
```

```python
top_student = df.loc[df["Marks"].idxmax()]

print(top_student)
```

```python
df.sort_values(by="Marks", ascending=False)
```

```python
plt.figure(figsize=(9, 5))

plt.bar(df["Student"], df["Marks"])

plt.title("Student Performance Analysis")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Marks", bins=5, kde=True)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    s=100
)

plt.title("Study Hours vs Student Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
correlation = df[
    ["Study_Hours", "Attendance", "Marks"]
].corr()

correlation
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.show()
```
